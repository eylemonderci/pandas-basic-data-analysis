```python
# İki boyutlu matrislere giriş

# 0-100 arasında rastgele sayılardan oluşan 3x2 boyutunda matris tanımlama
# 3 satır, 2 sütun => Eleman sayısı = 3*2 = 6

my_array_6 = np.random.randint(0,100,(3,2))
print(my_array_6)
```

```python
# Matrislerle ilgili sık kullanılan özellikler

print(f"Matrisin satır ve sütun sayısı = {my_array_6.shape}")
print("-"*50)
print(f"Matrisin satır sayısı = {my_array_6.shape[0]}")
print(f"Matrisin sütun sayısı = {my_array_6.shape[1]}")
print("-"*50)
print(f"Matrisin eleman sayısı = {my_array_6.size}")
print("-"*50)
print(f"Matrisin boyutu = {my_array_6.ndim}")
```

```python
my_array_7 = np.random.randint(0,100,(5,4))

print(f"Matrisle ilgili istatistiksel bilgiler")
print(f"En küçük eleman = {np.min(my_array_7)}")
print(f"En büyük eleman = {np.max(my_array_7)}")
print(f"Elemanların ortalama değeri = {np.mean(my_array_7)}")
print(f"En küçük elemanın indeks değeri = {np.argmin(my_array_7)}")
print(f"En büyük elemanın indeks değeri = {np.argmax(my_array_7)}")
print(f"Matrisin standart sapma değeri = {np.std(my_array_7)}")
print(f"Matrisin medyanı (ortanca değeri) = {np.median(my_array_7)}")
```

```python
# Özel matris üretmek

# İçi sadece sıfır (0) değeriyle dolu olan matris üretmek (istenilen boyutta)
my_array_8 = np.zeros((3,4))

# İçi sadece bir (0) değeriyle dolu olan matris üretmek (istenilen boyutta)
my_array_9 = np.ones((2,5))

# İçi tamamen sizin verdiğiniz değerlerle dolu olan matris (istenilen boyutta)
my_array_10 = np.full((3,3),12)

# Diyagonal matris (satır ve sütun sayısı eşit olan yerlerde 1, diğerlerinde)
# 0 değeri ile dolu olan matris
my_array_11 = np.eye(5)
```

```python
# Bir matris ile skalar bir değeri matematiksel işleme sokabilirsiniz
# yani bir matris ile bir int tipindeki değeri toplayabilirsiniz

my_array_12 = np.random.randint(1,10,(2,2))

print(my_array_12)
print("-"*50)
print(my_array_12+4)
```

```python
# her ikisi de 1 ile 100 değeri arasındaki rastgele değerlerden oluşan
# 3x3 matris tanımlayınız

my_array_13 = np.random.randint(0,100,(3,3))
my_array_14 = np.random.randint(0,100,(3,3))

print(my_array_13)
print("-"*50)
print(my_array_14)

# Bu iki matris arasında 4 temel matematiksel işlemi yapınız
print(f"Matrislerin Toplamı = {my_array_13+my_array_14}")
print("-"*50)
print(f"Matrislerin Farkı = {my_array_13-my_array_14}")
print("-"*50)
print(f"Matrislerin Çarpımı = {my_array_13*my_array_14}")
print("-"*50)
print(f"Matrislerin Bölümü = {my_array_13/my_array_14}")
print("-"*50)
# Ayrıca, bu matrislerin değerlerinin karesini alınız
print(f"1. Matrisin karesi = {my_array_13**2}")
print("-"*50)
print(f"2. Matrisin karesi = {my_array_14**2}")
print("-"*50)
# Son olarak her iki matristede yer alan tek sayıları gösteriniz
print(f"1. Matristeki Tek sayılar = {my_array_13%2!=0}")
print("-"*50)
print(f"2. Matristeki Tek sayılar = {my_array_14%2!=0}")
```

```python
# Bir numpy array içerisinde istenilen koşulu sağlayan elemanları bulma

my_array_15 = np.random.randint(0,100,(6,6))

# Array içerisindeki tek sayıların indeksleri
my_index_15 = np.where(my_array_15%2!=0)

# Array içerisindeki tek sayılar
my_values_15 = my_array_15[my_index_15]
#my_array_15[satır][sütun]

print(f"Orijinal Array = {my_array_15}")
print("-"*50)
print(f"İndeksler = {my_index_15}")
print("-"*50)
print(f"Değerler = {my_values_15}")
print(f"Toplam eleman = {my_values_15.size}")
```

```python
# 4x4 boyutunda 50 ile 100 arasında rastgele değerlerden oluşan bir matris
# tanımlayın
my_array_16 = np.random.randint(50,100,(4,4))

print(f"my_array_16 = {my_array_16}")
print("-"*50)
# Bu matristeki 75'ten büyük çift sayıları getiren kodu yazın
my_conditional_16 = (my_array_16>75) & (my_array_16%2==0)
my_index_16 = np.where(my_conditional_16)
print(f"Koşulu sağlayan değerler = {my_array_16[my_index_16]}")
```

```python
# Array'leri birleştirme

# 2 metot kullanılır
# tek boyutlu array olarak birleştirmek için concatenate

my_array_17 = np.array([1, 2, 3])
my_array_18 = np.array([11, 12, 13])
#deneme = np.array([21, 22, 23])
my_array_19_concat = np.concatenate((my_array_17, my_array_18))

# Array'leri olduğu gibi alt alta birleştirmek için stack
my_array_20_stack = np.stack((my_array_17, my_array_18))

print(f"Matris_1 = {my_array_17}")
print(f"Matris_2 = {my_array_18}")
print("-"*50)
print(f"Concat ile birleştirme = {my_array_19_concat}")
print("-"*50)
print(f"Stack ile birleştirme = {my_array_20_stack}")
```

```python
my_array_20 = np.array([1, 2, 2, -4, 5 ,98, 103, 16, -16, 32, 2, -4])
print(f"orijinal Array = {my_array_20}")
print("-"*20)
my_array_21_unique = np.unique(my_array_20)

print(f"Yeni Array = {my_array_21_unique}")
# yukarıdaki my_array_20'yi manuel olarak ( yani np.unique() kullanmadan)
# Önce tekrar eden değerlerden kurtulun
# sonra da bu değerleri sıralayın
# Son kertede sıralanmış verileri bir değişkene aktarın
# bu değişkeni ve tipini yazdırın (tipi numpy.ndarray olmalı!!!)
```

```python
# Aşama_1, tekrar eden verilerden kurtulma
my_set_1 = set(my_array_20)
print(f"Orijinal Array = {my_array_20}")
print(f"Yeni Değerler = {my_set_1}")
```

```python
# Aşama_2, Verilerin sıralı hale getirilmesi
my_array_sirali = np.array(sorted(my_set_1))
print(my_array_sirali)
print(type(my_array_sirali))
```

```python
# Array içerisinde istenilen aralıkta olmayan değerleri bizim verdiğimiz
# değerlerle değiştiren metot

my_array_22 = np.array([15,38,4500,180,-58,0,29,45,49,1,18,-1200])
print(f"Orijinal Array = {my_array_22}")
print("-"*50)

my_array_23_clip = np.clip(my_array_22,20,100)
print(f"İşlenmiş Array = {my_array_23_clip}")

# Alt limit 40
# Üst limit array'in ortalama değeri

my_array_24_clip = np.clip(my_array_22,40,np.mean(my_array_22))
print("-"*50)
print(f"Yeni İşlenmiş Array (mean) = {my_array_24_clip}")
```
